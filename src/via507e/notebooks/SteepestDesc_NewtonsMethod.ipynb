```python
import numpy as np
```

```python
def f(x):
    return (x[0]-2)**4+np.exp(x[0]-2)+(x[0]-2*x[1])**2
```

```python
x=np.array([0,3])
```

```python
f(x)
```

```python
def gradient_f(x):
    dfdx1=4*(x[0]-2)**3+np.exp(x[0]-2)+2*(x[0]-2*x[1])
    dfdx2=-4*(x[0]-2*x[1])
    return np.array([dfdx1,dfdx2])
```

```python
gradient_f(x)
```

```python
def hessian_f(x):
    dfdx1dx1=12*(x[0]-2)**2+np.exp(x[0]-2)+2
    dfdx1dx2=-4
    dfdx2dx1=-4
    dfdx2dx2=8
    return np.array([[dfdx1dx1,dfdx1dx2],[dfdx2dx1,dfdx2dx2]])
```

```python
x=np.array([0,3])
```

```python
hessian_f(x)
```

```python
#Steepest Descent with alpha=0.01
delta=1000
alpha=0.001
x=[0,3]
iteration=0
while delta>0.000001:
    iteration=iteration+1
    old_x=x
    x=x-alpha*gradient_f(x)
    delta=np.sqrt(np.sum((old_x-x)**2))
    print("At iteration: ", iteration, " x values are : ",x, " with a value of f(x): ", f(x))
```

```python
#Newton's Method
delta=1000
alpha=0.01
x=np.array([0,3])
iteration=0
while delta>0.000001:
    iteration=iteration+1
    old_x=x
    x=x-np.dot(np.linalg.inv(hessian_f(x)),gradient_f(x))
    delta=np.sqrt(np.sum((old_x-x)**2))
    print("At iteration: ", iteration, " x values are : ",x, " with a value of f(x): ", f(x))
```
